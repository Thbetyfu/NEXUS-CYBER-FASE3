# NEX-AI Colab Workflow

Notebook ini disiapkan untuk workflow **IDE + Google Colab extension**.

Tujuan utamanya:
- mount Google Drive,
- verifikasi struktur `MyDrive/NEX-AI-Collab`,
- install dependency,
- menjalankan `train_qlora.py` yang ada di Google Drive,
- mengecek hasil `checkpoints` dan menyiapkan zip artefak.

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
from pathlib import Path

PROJECT_DIR_NAME = 'NEX-AI-Collab'
DRIVE_ROOT = Path('/content/drive/MyDrive')
BASE_DIR = DRIVE_ROOT / PROJECT_DIR_NAME
DATASET_PATH = BASE_DIR / 'dataset' / 'cyber_security_dataset.json'
SCRIPT_PATH = BASE_DIR / 'scripts' / 'train_qlora.py'
CHECKPOINT_DIR = BASE_DIR / 'checkpoints'
OUTPUTS_DIR = BASE_DIR / 'outputs'

paths = {
    'BASE_DIR': BASE_DIR,
    'DATASET_PATH': DATASET_PATH,
    'SCRIPT_PATH': SCRIPT_PATH,
    'CHECKPOINT_DIR': CHECKPOINT_DIR,
    'OUTPUTS_DIR': OUTPUTS_DIR,
}

for name, path in paths.items():
    print(f'{name}: {path} -> {path.exists()}')
```

```python
import torch

print('CUDA available:', torch.cuda.is_available())
if not torch.cuda.is_available():
    raise RuntimeError('GPU Colab belum aktif. Gunakan Runtime -> Change runtime type -> GPU.')

print('GPU:', torch.cuda.get_device_name(0))
print('VRAM (GiB):', round(torch.cuda.get_device_properties(0).total_memory / (1024 ** 3), 2))
```
